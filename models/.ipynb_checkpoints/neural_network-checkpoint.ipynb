```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split

from tensorflow.keras import Sequential, Input
from tensorflow.keras.layers import Dense, Dropout
from tensorflow.keras.utils import to_categorical
from tensorflow.keras import regularizers
from tensorflow.keras.optimizers import Adam
```

```python
df = pd.read_parquet("../data/cookie_cats/clean.parquet")
# Convert type (catergoy) to int
df["version"] = df["version"].astype(str).map({"gate_30": 0, "gate_40": 1}).astype(int)
print(df.dtypes)
print(df.head())
```

```python
# Get labels & values
X = df[["version", "sum_gamerounds"]]

y_retention_1 = df["retention_1"]
y_retention_7 = df["retention_7"]

z_score = StandardScaler()
X = z_score.fit_transform(X) 
```

```python
# Define Model Architecture - One Day Retention
model_1 = Sequential([
    Input(shape=(X.shape[1],)), 
    Dense(128, activation='relu'),
    Dense(64,  activation='relu'),
    Dense(1,  activation='sigmoid')
])

optimizer = Adam(learning_rate=0.001)
model_1.compile(optimizer=optimizer, loss='binary_crossentropy', metrics=['accuracy'])
```

```python
# Predict One Day Retention
X1_train, X1_test, y1_train, y1_test = train_test_split(X, y_retention_1, test_size=0.2, random_state=1166, stratify=y_retention_1)

train_model_1 = model_1.fit(
    X1_train, y1_train,
    epochs=30,
    batch_size=32,
    validation_data=(X1_test, y1_test)
)
```

```python
loss, accuracy = model_1.evaluate(X1_test, y1_test, verbose=0)
print(f"Test accuracy: {accuracy:.4f}  Loss:{loss:.4f}")

# --- Plot 1: Training vs Validation Accuracy ---
plt.figure(figsize=(8,5))
plt.plot(train_model_1.history['accuracy'], label='Training Accuracy')
plt.plot(train_model_1.history['val_accuracy'], label='Validation Accuracy')
plt.title('Training vs Validation Accuracy')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()
plt.grid(True)
plt.show()

# --- Plot 2: Training vs Validation Loss ---
plt.figure(figsize=(8,5))
plt.plot(train_model_1.history['loss'], label='Training Loss')
plt.plot(train_model_1.history['val_loss'], label='Validation Loss')
plt.title('Training vs Validation Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Define Model Architecture - Seven Day Retention
model_7 = Sequential([
    Input(shape=(X.shape[1],)), 
    Dense(128, activation='relu'),
    Dense(64,  activation='relu'),
    Dense(1,  activation='sigmoid')
])

optimizer = Adam(learning_rate=0.001)
model_7.compile(optimizer=optimizer, loss='binary_crossentropy', metrics=['accuracy'])
```

```python
# Predict Seven Day Retention
X7_train, X7_test, y7_train, y7_test = train_test_split(X, y_retention_7, test_size=0.2, random_state=1166, stratify=y_retention_7)

train_model_7 = model_7.fit(
    X7_train, y7_train,
    epochs=30,
    batch_size=32,
    validation_data=(X7_test, y7_test)
)
```

```python
loss, accuracy = model_1.evaluate(X1_test, y1_test, verbose=0)
print(f"Test accuracy: {accuracy:.4f}  Loss:{loss:.4f}")

# --- Plot 1: Training vs Validation Accuracy ---
plt.figure(figsize=(8,5))
plt.plot(train_model_7.history['accuracy'], label='Training Accuracy')
plt.plot(train_model_7.history['val_accuracy'], label='Validation Accuracy')
plt.title('Training vs Validation Accuracy')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()
plt.grid(True)
plt.show()

# --- Plot 2: Training vs Validation Loss ---
plt.figure(figsize=(8,5))
plt.plot(train_model_7.history['loss'], label='Training Loss')
plt.plot(train_model_7.history['val_loss'], label='Validation Loss')
plt.title('Training vs Validation Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()
plt.grid(True)
plt.show()
```
