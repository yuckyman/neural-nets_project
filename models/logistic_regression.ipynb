```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

```python
df = pd.read_parquet("../data/cookie_cats/clean.parquet")
```

```python
# Convert type (catergoy) to int
df["version"] = df["version"].astype(str).map({"gate_30": 0, "gate_40": 1}).astype(int)
print(df.dtypes)
print(df.head())
```

```python
X = df[["version", "sum_gamerounds"]]
```

```python
# Model - One day retention prediction
y_retention_1 = df["retention_1"]   # 1 day retention

X1_train, X1_test, y1_train, y1_test = train_test_split(X, y_retention_1, test_size=0.2, random_state=1166)
model_retention_1 = LogisticRegression().fit(X1_train, y1_train)

pred_retention_1 = model_retention_1.predict(X1_test)

print("Day-1 Retention")
print(f"Accuracy: {accuracy_score(y1_test, pred_retention_1):.4f}")
print(classification_report(y1_test, model_retention_1.predict(X1_test)),"\n")
```

```python
# Model - Seven day retention
y_retention_7 = df["retention_7"]   # 7 day retention

X7_train, X7_test, y7_train, y7_test = train_test_split(X, y_retention_7, test_size=0.2, random_state=1166)
model_retention_7 = LogisticRegression().fit(X7_train, y7_train)

pred_retention_7 = model_retention_7.predict(X7_test)

print("Day-7 Retention")
print(f"Accuracy: {accuracy_score(y7_test, pred_retention_7):.4f}")
print(classification_report(y7_test, model_retention_7.predict(X7_test)),"\n")
```
